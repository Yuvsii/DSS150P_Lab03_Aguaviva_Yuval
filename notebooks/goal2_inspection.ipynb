# Goal 2 Inspection

```
import pandas as pd
from pathlib import Path

raw_dir = Path('../data/raw')
staging_dir = Path('../data/staging')
curated_dir = Path('../data/curated')
quarantine_dir = Path('../data/quarantine')
```

## 1. Raw Data

```
last_run_dir = sorted(raw_dir.glob('run_id=*'))[-1]
print('Latest run:', last_run_dir)

orders_raw = pd.read_csv(last_run_dir / 'orders.csv')
display(orders_raw.head())
```

## 2. Staging Data

```
orders_staging = pd.read_parquet(staging_dir / 'orders.parquet')
print(f'Staging orders rows: {len(orders_staging)}')
display(orders_staging.head())
```

## 3. Curated Data

```
curated = pd.read_parquet(curated_dir / 'sales_order_lines.parquet')
print(f'Curated rows: {len(curated)}')
display(curated.head())
```

## 4. Quarantine Data

```
last_q_file = sorted(quarantine_dir.glob('quarantine_run_*.csv'))[-1]
q_staging = pd.read_csv(last_q_file)
print(f'Quarantine (Staging) rows: {len(q_staging)}')
display(q_staging.head())

last_q_curated_file = sorted(quarantine_dir.glob('quarantine_curated_*.csv'))[-1]
q_curated = pd.read_csv(last_q_curated_file)
print(f'Quarantine (Curated) rows: {len(q_curated)}')
display(q_curated.head())
```
